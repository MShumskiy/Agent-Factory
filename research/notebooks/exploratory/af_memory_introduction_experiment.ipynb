```python
import os
import psycopg2 as psql
from pydantic import BaseModel
from typing import  List
import sys
import os
import json


db_database = "llmp_db"
db_user = os.getenv("DB_USER")
db_password = os.getenv("DB_PASSWORD")
db_host = os.getenv("DB_HOST")

DB_CONFIG = {
    "dbname": db_database,
    "user": db_user,
    "password": db_password,
    "host": db_host
}


config_file_path ="configs/rag_configs.json"

        
with open(config_file_path, 'r') as config_file:
        config = json.load(config_file)
        
system_prompt = config['system_prompt_rag']
embeddings_model_id = config['embeddings_model_id']
cross_encoder_id = config['cross_encoder_id']
top_k = config['top_k']
temperature = config['temperature']
ce_threshold = config['ce_threshold']
search_type = config['search_type']
src = config['src']
```

```python
import pandas as pd
def query_gen_hist_db():
    """
    Retrieves all rows from the generation_history_v4 table as a DataFrame.
    """
    conn = psql.connect(**DB_CONFIG)
    query = "SELECT * FROM generation_history_v4;"
    
    df = pd.read_sql_query(query, conn)
    conn.close()
    
    return df
```
